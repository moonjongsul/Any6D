```python
# SAM을 사용한 물체 Annotation
import torch
import numpy as np
import cv2
import matplotlib.pyplot as plt
from PIL import Image
import os

# SAM2 import
from sam2.sam2.build_sam import build_sam2
from sam2.sam2.sam2_image_predictor import SAM2ImagePredictor
```

```python
# SAM2 모델 초기화
def init_sam2():
    """SAM2 모델을 초기화하고 predictor를 반환"""
    # SAM2 설정
    sam2_checkpoint = "sam2/checkpoints/sam2.1_hiera_large.pt"
    model_cfg = "sam2/configs/sam2.1/sam2.1_hiera_l.yaml"
    device = "cuda" if torch.cuda.is_available() else "cpu"
    
    # 모델 빌드
    sam2_model = build_sam2(model_cfg, sam2_checkpoint, device=device)
    
    # Predictor 생성
    predictor = SAM2ImagePredictor(sam2_model)
    
    return predictor

# 모델 초기화
print("SAM2 모델 로딩 중...")
predictor = init_sam2()
print("SAM2 모델 로딩 완료!")
```

```python
# 이미지 로드 및 전처리
def load_and_preprocess_image(image_path):
    """이미지를 로드하고 RGB로 변환"""
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"이미지를 로드할 수 없습니다: {image_path}")
    
    # BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    return image_rgb
```

```python

# 이미지 로드
image = load_and_preprocess_image(color_path)
print(f"이미지 크기: {image.shape}")

# 이미지 표시
plt.figure(figsize=(10, 8))
plt.imshow(image)
plt.title("raw")
plt.axis('off')
plt.show()
```

```python
# SAM2를 사용한 물체 Annotation 함수들
def annotate_with_points(predictor, image, points, labels):
    """
    점을 사용한 물체 annotation
    
    Args:
        predictor: SAM2 predictor
        image: 입력 이미지 (RGB)
        points: 클릭한 점들의 좌표 [(x1, y1), (x2, y2), ...]
        labels: 각 점의 라벨 [1, 1, 0, ...] (1: foreground, 0: background)
    """
    # 이미지 설정
    predictor.set_image(image)
    
    # 점들을 numpy 배열로 변환
    input_points = np.array(points)
    input_labels = np.array(labels)
    
    # 마스크 생성
    masks, scores, logits = predictor.predict(
        point_coords=input_points,
        point_labels=input_labels,
        multimask_output=True,
    )
    
    return masks, scores, logits

def annotate_with_box(predictor, image, box):
    """
    박스를 사용한 물체 annotation
    
    Args:
        predictor: SAM2 predictor
        image: 입력 이미지 (RGB)
        box: [x1, y1, x2, y2] 형태의 박스 좌표
    """
    # 이미지 설정
    predictor.set_image(image)
    
    # 박스를 numpy 배열로 변환
    input_box = np.array(box)
    
    # 마스크 생성
    masks, scores, logits = predictor.predict(
        point_coords=None,
        point_labels=None,
        box=input_box[None, :],
        multimask_output=False,
    )
    
    return masks, scores, logits

def visualize_annotation(image, masks, scores, points=None, labels=None, box=None):
    """Annotation 결과를 시각화"""
    num_masks = len(masks)
    
    fig, axes = plt.subplots(2, (num_masks + 1) // 2, figsize=(15, 10))
    if num_masks == 1:
        axes = [axes]
    else:
        axes = axes.flatten()
    
    # 원본 이미지
    axes[0].imshow(image)
    axes[0].set_title("원본 이미지")
    axes[0].axis('off')
    
    # 점 표시
    if points is not None:
        for i, (point, label) in enumerate(zip(points, labels)):
            color = 'red' if label == 1 else 'blue'
            axes[0].plot(point[0], point[1], 'o', color=color, markersize=10)
            axes[0].text(point[0], point[1], f'{i+1}', color='white', fontsize=12, ha='center', va='center')
    
    # 박스 표시
    if box is not None:
        x1, y1, x2, y2 = box
        rect = plt.Rectangle((x1, y1), x2-x1, y2-y1, linewidth=2, edgecolor='green', facecolor='none')
        axes[0].add_patch(rect)
    
    # 각 마스크 표시
    for i, (mask, score) in enumerate(zip(masks, scores)):
        if i + 1 < len(axes):
            axes[i + 1].imshow(image)
            axes[i + 1].imshow(mask, alpha=0.6, cmap='Reds')
            axes[i + 1].set_title(f"마스크 {i+1} (점수: {score:.3f})")
            axes[i + 1].axis('off')
    
    # 빈 subplot 숨기기
    for i in range(num_masks + 1, len(axes)):
        axes[i].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    return masks, scores
```

```python
# 방법 1: 점을 사용한 Annotation
print("=== 방법 1: 점을 사용한 Annotation ===")
print("이미지에서 물체의 중심 부분을 클릭하세요.")

# 예시: 물체의 중심 부분에 점 설정 (이미지 크기에 맞게 조정)
h, w = image.shape[:2]
center_x, center_y = w // 2, h // 2

# 물체가 있는 부분에 점 설정 (예시 좌표)
points = [
    (center_x, center_y),  # 물체 중심
    (center_x + 50, center_y + 30),  # 물체 내부 다른 점
]

# 각 점의 라벨 (1: foreground, 0: background)
labels = [1, 1]  # 모두 물체 내부 점

print(f"설정된 점들: {points}")
print(f"점 라벨들: {labels}")

# Annotation 실행
masks, scores, logits = annotate_with_points(predictor, image, points, labels)

# 결과 시각화
visualize_annotation(image, masks, scores, points=points, labels=labels)
```

```python
# 방법 2: 박스를 사용한 Annotation
print("=== 방법 2: 박스를 사용한 Annotation ===")
print("물체를 둘러싸는 박스를 설정하세요.")

# 예시: 물체를 둘러싸는 박스 설정 (이미지 크기에 맞게 조정)
box = [
    center_x - 100,  # x1
    center_y - 100,  # y1
    center_x + 100,  # x2
    center_y + 100   # y2
]

print(f"설정된 박스: {box}")

# Annotation 실행
masks_box, scores_box, logits_box = annotate_with_box(predictor, image, box)

# 결과 시각화
visualize_annotation(image, masks_box, scores_box, box=box)
```

```python
# 방법 3: 점 + 박스 조합 Annotation
print("=== 방법 3: 점 + 박스 조합 Annotation ===")
print("점과 박스를 함께 사용하여 더 정확한 annotation을 수행합니다.")

def annotate_with_points_and_box(predictor, image, points, labels, box):
    """점과 박스를 함께 사용한 annotation"""
    predictor.set_image(image)
    
    input_points = np.array(points)
    input_labels = np.array(labels)
    input_box = np.array(box)
    
    masks, scores, logits = predictor.predict(
        point_coords=input_points,
        point_labels=input_labels,
        box=input_box[None, :],
        multimask_output=True,
    )
    
    return masks, scores, logits

# 점과 박스 조합으로 annotation
masks_combined, scores_combined, logits_combined = annotate_with_points_and_box(
    predictor, image, points, labels, box
)

# 결과 시각화
visualize_annotation(image, masks_combined, scores_combined, 
                    points=points, labels=labels, box=box)
```

```python
# 마스크 저장 및 후처리
def save_masks(masks, scores, save_dir="./sam_annotations"):
    """생성된 마스크들을 저장"""
    os.makedirs(save_dir, exist_ok=True)
    
    saved_files = []
    for i, (mask, score) in enumerate(zip(masks, scores)):
        # 마스크를 0-255 범위로 변환
        mask_uint8 = (mask * 255).astype(np.uint8)
        
        # 파일명 생성
        filename = f"mask_{i+1}_score_{score:.3f}.png"
        filepath = os.path.join(save_dir, filename)
        
        # 저장
        cv2.imwrite(filepath, mask_uint8)
        saved_files.append(filepath)
        print(f"마스크 {i+1} 저장: {filepath} (점수: {score:.3f})")
    
    return saved_files

# 최고 점수 마스크 선택
best_mask_idx = np.argmax(scores_combined)
best_mask = masks_combined[best_mask_idx]
best_score = scores_combined[best_mask_idx]

print(f"최고 점수 마스크: {best_mask_idx + 1} (점수: {best_score:.3f})")

# 마스크들 저장
saved_files = save_masks(masks_combined, scores_combined)

# 최고 점수 마스크 시각화
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.imshow(best_mask, alpha=0.6, cmap='Reds')
plt.title(f"최고 점수 마스크 (점수: {best_score:.3f})")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(best_mask, cmap='gray')
plt.title("마스크만 표시")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# 인터랙티브 Annotation을 위한 유틸리티 함수
def interactive_annotation():
    """인터랙티브하게 점과 박스를 설정할 수 있는 함수"""
    print("=== 인터랙티브 Annotation ===")
    print("다음 중 하나를 선택하세요:")
    print("1. 점으로 annotation")
    print("2. 박스로 annotation") 
    print("3. 점 + 박스 조합 annotation")
    
    # 사용자 입력 받기 (실제로는 마우스 클릭으로 구현 가능)
    choice = input("선택 (1/2/3): ")
    
    if choice == "1":
        print("물체 내부에 점을 클릭하세요 (좌표를 입력하세요)")
        points = []
        labels = []
        
        while True:
            try:
                x = int(input("x 좌표 (종료하려면 -1): "))
                if x == -1:
                    break
                y = int(input("y 좌표: "))
                label = int(input("라벨 (1: 물체, 0: 배경): "))
                
                points.append((x, y))
                labels.append(label)
                print(f"점 추가: ({x}, {y}) - 라벨: {label}")
            except ValueError:
                print("올바른 숫자를 입력하세요.")
        
        if points:
            masks, scores, _ = annotate_with_points(predictor, image, points, labels)
            visualize_annotation(image, masks, scores, points=points, labels=labels)
            return masks, scores
    
    elif choice == "2":
        print("물체를 둘러싸는 박스를 설정하세요")
        try:
            x1 = int(input("왼쪽 위 x: "))
            y1 = int(input("왼쪽 위 y: "))
            x2 = int(input("오른쪽 아래 x: "))
            y2 = int(input("오른쪽 아래 y: "))
            
            box = [x1, y1, x2, y2]
            masks, scores, _ = annotate_with_box(predictor, image, box)
            visualize_annotation(image, masks, scores, box=box)
            return masks, scores
        except ValueError:
            print("올바른 숫자를 입력하세요.")
    
    elif choice == "3":
        print("점과 박스를 모두 설정하세요")
        # 점 설정
        points = []
        labels = []
        print("점 설정 (종료하려면 x 좌표에 -1 입력)")
        while True:
            try:
                x = int(input("x 좌표: "))
                if x == -1:
                    break
                y = int(input("y 좌표: "))
                label = int(input("라벨 (1: 물체, 0: 배경): "))
                points.append((x, y))
                labels.append(label)
            except ValueError:
                print("올바른 숫자를 입력하세요.")
        
        # 박스 설정
        print("박스 설정")
        try:
            x1 = int(input("왼쪽 위 x: "))
            y1 = int(input("왼쪽 위 y: "))
            x2 = int(input("오른쪽 아래 x: "))
            y2 = int(input("오른쪽 아래 y: "))
            box = [x1, y1, x2, y2]
            
            masks, scores, _ = annotate_with_points_and_box(predictor, image, points, labels, box)
            visualize_annotation(image, masks, scores, points=points, labels=labels, box=box)
            return masks, scores
        except ValueError:
            print("올바른 숫자를 입력하세요.")
    
    return None, None

# 인터랙티브 annotation 실행 (주석 해제하여 사용)
# masks_interactive, scores_interactive = interactive_annotation()
```
